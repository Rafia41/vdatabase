```python
## this is for gemnin

import os

import numpy as np
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

EMBEDDING_DIM = 768  # choose 768, 1536 or 3072


def get_embedding(sentence, task_type="SEMANTIC_SIMILARITY"):
    result = client.models.embed_content(
        model="gemini-embedding-001",
        contents=sentence,
        config=types.EmbedContentConfig(
            task_type=task_type,
            output_dimensionality=EMBEDDING_DIM,
        ),
    )
    vec = np.array(result.embeddings[0].values)
    return (vec / np.linalg.norm(vec)).tolist()  # normalize
```

```python
import os
import numpy as np

if os.path.exists("embedded_sentences_.csv"):
    dataset = pd.read_csv("embedded_sentences.csv")
    dataset["embedding"] = dataset.embedding.apply(eval).apply(np.array)
else:
    dataset["embedding"] = dataset["sentence"].apply(get_embedding)
    dataset.to_csv("embedded_sentences_fia.csv", index=False)
```

```python
embeddings = np.array(dataset.embedding.tolist())

query = "A man is playing a flute"
xq = get_embedding(query)
```

```python
def search(query, k=5):
    xq = np.array(get_embedding(query))
    scores = embeddings @ xq
    top_idx = np.argsort(scores)[::-1][:k]
    return dataset.iloc[top_idx].assign(score=scores[top_idx])[["id", "sentence", "score"]]


search("A girl is doing her makeup",5)
```

## Faiss

Check [Faiss Indexes](https://github.com/facebookresearch/faiss/wiki/Faiss-indexes) for more information.


```python
embeddings = np.array(dataset.embedding.tolist())

query = "A girl is playing"
xq = get_embedding(query)
```

### IndexFlatL2 - Exact Search for L2


```python
import faiss

index_l2 = faiss.IndexFlatL2(embedding_dimension)
index_l2.is_trained
```

```python
index_l2.add(embeddings)
index_l2.ntotal
```

```python
_, document_indices = index_l2.search(np.expand_dims(xq, axis=0), k=4)
dataset.iloc[document_indices[0]]
```

### IndexIVFFlat - Inverted file with exact post-verification

<img src='images/ivf.png' width="1000">


```python
ncentroids = 20
quantizer = faiss.IndexFlatL2(embedding_dimension)
index_ivf = faiss.IndexIVFFlat(quantizer, embedding_dimension, ncentroids)
index_ivf.is_trained
```

```python
index_ivf.train(embeddings)
index_ivf.is_trained
```

```python
index_ivf.add(embeddings)
index_ivf.ntotal
```

```python
_, document_indices = index_ivf.search(np.expand_dims(xq, axis=0), k=4)
dataset.iloc[document_indices[0]]
```

```python
index_ivf.nprobe = 5
_, document_indices = index_ivf.search(np.expand_dims(xq, axis=0), k=4)
dataset.iloc[document_indices[0]]
```

### IndexIVFPQ - IVF + Product Quantizer (PQ)

<img src='images/ivf-pq.png' width="1000">

```python
code_size = 8
bits_per_centroid = 4

index_ivf_pq = faiss.IndexIVFPQ(
    quantizer, embedding_dimension, ncentroids, code_size, bits_per_centroid
)
index_ivf_pq.is_trained
```

```python
index_ivf_pq.train(embeddings)
index_ivf_pq.add(embeddings)
index_ivf_pq.ntotal
```

```python
index_ivf_pq.nprobe = 5
_, document_indices = index_ivf_pq.search(np.expand_dims(xq, axis=0), k=4)
dataset.iloc[document_indices[0]]
```

```python
import os
import time
import ast
import pandas as pd
from dotenv import load_dotenv
from pinecone import Pinecone, ServerlessSpec

load_dotenv(override=True)

# ---------- 1. Connect ----------
api_key = os.getenv("PINECONE_API_KEY")
print("key starts with:", api_key[:8])        # compare with the key in the console's API keys page
database = Pinecone(api_key=api_key)

# ---------- 2. Load data ----------
dataset = pd.read_csv("embedded_sentences_fia.csv")   # columns: sentence, embedding
dataset["id"] = range(1, len(dataset) + 1)

def parse_embedding(x):
    if isinstance(x, str):
        x = x.strip()
        if x.startswith("["):
            return [float(v) for v in ast.literal_eval(x)]
        return [float(v) for v in x.replace(",", " ").split()]
    return [float(v) for v in x]

dataset["embedding"] = dataset["embedding"].apply(parse_embedding)
embedding_dimension = len(dataset["embedding"].iloc[0])
print("rows:", len(dataset), "| dimension:", embedding_dimension)

# ---------- 3. Create the vector index ----------
INDEX_NAME = "databasevector"
if INDEX_NAME in database.list_indexes().names():
    database.delete_index(INDEX_NAME)
    time.sleep(5)

database.create_index(
    name="databasevector",
    dimension=768,
    metric="cosine",
    spec=ServerlessSpec(cloud="aws", region="us-east-1"),
)
while not database.describe_index("databasevector").status["ready"]:
    time.sleep(1)

pinecone_index = database.Index(INDEX_NAME)
print("host:", database.describe_index(INDEX_NAME).host)   # compare with the host in the console

# ---------- 4. Helpers ----------
def iterator(dataset, size):
    for i in range(0, len(dataset), size):
        yield dataset.iloc[i : i + size]

def vector(batch):
    vectors = []
    for _, row in batch.iterrows():
        vectors.append({
            "id": str(row["id"]),
            "values": [float(v) for v in row["embedding"]],
            "metadata": {"sentence": str(row["sentence"])},
        })
    return vectors

# ---------- 5. Upsert ----------
for batch in iterator(dataset, 100):
    pinecone_index.upsert(vectors=vector(batch))

time.sleep(5)
print(pinecone_index.describe_index_stats())
```

```python
import os
from dotenv import load_dotenv
from pinecone import Pinecone
from sentence_transformers import SentenceTransformer

load_dotenv(override=True)

database = Pinecone(api_key=os.getenv("PINECONE_API_KEY"))
pinecone_index = database.Index("databasevector")     # connects to existing data, no recreate
model = SentenceTransformer("all-mpnet-base-v2")      # loads from cache

def search(text, k=3):
    emb = model.encode(text).tolist()
    res = pinecone_index.query(vector=emb, top_k=k, include_metadata=True)
    for m in res["matches"]:
        print(round(m["score"], 3), "|", m["metadata"]["sentence"])

print("ready:", pinecone_index.describe_index_stats()["total_vector_count"], "vectors")
```
